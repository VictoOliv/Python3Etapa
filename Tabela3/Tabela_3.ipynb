```python
bruto = pd.read_excel(path, sheet_name="Tabela base do SIDRA 10063", header=None)
bruto
```

```python
# Dados por Brasil, regiões e UFs (sem a linha de fonte)
dados = bruto.iloc[8:41].copy()
dados
```

## Divisão em 3 tabelas

1. **Curso superior** — total de pessoas com graduação concluída
2. **Ciência, Tecnologia, Engenharias e Matemática** (STEM)
3. **Educação, Serviços pessoais, Saúde e Bem-estar**

```python
superior_total = dados.iloc[:, [0, 1, 2, 3]].copy()
superior_total.columns = ["regiao", "total", "homens", "mulheres"]
superior_total = superior_total.reset_index(drop=True)

for col in superior_total.columns[1:]:
    superior_total[col] = pd.to_numeric(superior_total[col], errors="coerce")

superior_total
```

```python
stem = dados.iloc[:, [0, 27, 28, 29]].copy()
stem.columns = ["regiao", "total", "homens", "mulheres"]
stem = stem.reset_index(drop=True)

for col in stem.columns[1:]:
    stem[col] = pd.to_numeric(stem[col], errors="coerce")

stem
```

```python
educacao_servicos_saude = dados.iloc[:, [0, 30, 31, 32]].copy()
educacao_servicos_saude.columns = ["regiao", "total", "homens", "mulheres"]
educacao_servicos_saude = educacao_servicos_saude.reset_index(drop=True)

for col in educacao_servicos_saude.columns[1:]:
    educacao_servicos_saude[col] = pd.to_numeric(educacao_servicos_saude[col], errors="coerce")

educacao_servicos_saude
```
